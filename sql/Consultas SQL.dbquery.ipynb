```sql
SHOW CATALOGS;

SHOW SCHEMAS IN fia_databricks;

SHOW TABLES IN fia_databricks.datos;

SHOW VOLUMES IN fia_databricks.datos;



SELECT
    COUNT(*) AS total_registros
FROM fia_databricks.datos.renipress_silver;



SELECT
    DEPARTAMENTO,
    COUNT(*) AS total_ipress
FROM fia_databricks.datos.renipress_silver
GROUP BY DEPARTAMENTO
ORDER BY total_ipress DESC;



SELECT
    CATEGORIA,
    COUNT(*) AS total_ipress
FROM fia_databricks.datos.renipress_silver
GROUP BY CATEGORIA
ORDER BY total_ipress DESC;



SELECT
    DEPARTAMENTO,
    CATEGORIA,
    COUNT(*) AS total_ipress
FROM fia_databricks.datos.renipress_silver
GROUP BY
    DEPARTAMENTO,
    CATEGORIA
ORDER BY
    DEPARTAMENTO,
    total_ipress DESC;



SELECT
    COD_IPRESS,
    COUNT(*) AS cantidad
FROM fia_databricks.datos.renipress_silver
WHERE COD_IPRESS IS NOT NULL
GROUP BY COD_IPRESS
HAVING COUNT(*) > 1
ORDER BY cantidad DESC;



SELECT
    COD_IPRESS,
    NOMBRE,
    DEPARTAMENTO,
    CATEGORIA
FROM fia_databricks.datos.renipress_silver
WHERE
    COD_IPRESS IS NULL
    OR NOMBRE IS NULL
    OR DEPARTAMENTO IS NULL
    OR CATEGORIA IS NULL
LIMIT 100;


```
