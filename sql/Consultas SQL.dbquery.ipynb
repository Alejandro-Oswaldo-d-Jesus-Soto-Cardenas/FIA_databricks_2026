```sql
-- =====================================================
-- PROYECTO: STASYS
-- TEAM: 02
-- ALUMNO: ALEJANDRO
-- PLATAFORMA: DATABRICKS
-- DATASET: RENIPRESS
-- =====================================================


-- 1. VERIFICAR LOS CATÁLOGOS
SHOW CATALOGS;


-- 2. CREAR EL ESQUEMA DEL PROYECTO
CREATE SCHEMA IF NOT EXISTS
fia_databricks.stasys_team02_alejandro;


-- 3. VERIFICAR LOS ESQUEMAS
SHOW SCHEMAS IN fia_databricks;


-- 4. VERIFICAR LAS TABLAS DEL ESQUEMA
SHOW TABLES IN fia_databricks.stasys_team02_alejandro;


-- 5. VERIFICAR LOS VOLÚMENES DEL ESQUEMA ORIGINAL
-- Se conserva datos porque ahí está fia_volume.
SHOW VOLUMES IN fia_databricks.datos;


-- =====================================================
-- CONSULTAS SQL SOBRE RENIPRESS
-- Requieren que renipress_silver ya esté cargada
-- en fia_databricks.stasys_team02_alejandro.
-- =====================================================


-- 6. TOTAL DE REGISTROS
SELECT
    COUNT(*) AS total_registros
FROM fia_databricks.stasys_team02_alejandro.renipress_silver;


-- 7. CANTIDAD DE IPRESS POR DEPARTAMENTO
SELECT
    COALESCE(DEPARTAMENTO, 'SIN DEPARTAMENTO') AS departamento,
    COUNT(*) AS total_ipress
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
GROUP BY COALESCE(DEPARTAMENTO, 'SIN DEPARTAMENTO')
ORDER BY total_ipress DESC;


-- 8. CANTIDAD DE IPRESS POR CATEGORÍA
SELECT
    COALESCE(CATEGORIA, 'SIN CATEGORIA') AS categoria,
    COUNT(*) AS total_ipress
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
GROUP BY COALESCE(CATEGORIA, 'SIN CATEGORIA')
ORDER BY total_ipress DESC;


-- 9. CANTIDAD DE IPRESS POR DEPARTAMENTO Y CATEGORÍA
SELECT
    COALESCE(DEPARTAMENTO, 'SIN DEPARTAMENTO') AS departamento,
    COALESCE(CATEGORIA, 'SIN CATEGORIA') AS categoria,
    COUNT(*) AS total_ipress
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
GROUP BY
    COALESCE(DEPARTAMENTO, 'SIN DEPARTAMENTO'),
    COALESCE(CATEGORIA, 'SIN CATEGORIA')
ORDER BY
    departamento,
    total_ipress DESC;


-- 10. BUSCAR CÓDIGOS IPRESS DUPLICADOS
SELECT
    COD_IPRESS,
    COUNT(*) AS cantidad
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE COD_IPRESS IS NOT NULL
  AND TRIM(COD_IPRESS) <> ''
GROUP BY COD_IPRESS
HAVING COUNT(*) > 1
ORDER BY cantidad DESC;


-- 11. MOSTRAR REGISTROS CON DATOS INCOMPLETOS
-- Considera valores nulos y cadenas vacías.
SELECT
    COD_IPRESS,
    NOMBRE,
    DEPARTAMENTO,
    CATEGORIA,
    INSTITUCION
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE
       COD_IPRESS IS NULL OR TRIM(COD_IPRESS) = ''
    OR NOMBRE IS NULL OR TRIM(NOMBRE) = ''
    OR DEPARTAMENTO IS NULL OR TRIM(DEPARTAMENTO) = ''
    OR CATEGORIA IS NULL OR TRIM(CATEGORIA) = ''
LIMIT 100;


-- 12. CONTAR REGISTROS CON DATOS INCOMPLETOS
-- Cuenta cada registro una sola vez, aunque tenga
-- más de un campo incompleto.
SELECT
    COUNT(*) AS registros_incompletos
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE
       COD_IPRESS IS NULL OR TRIM(COD_IPRESS) = ''
    OR NOMBRE IS NULL OR TRIM(NOMBRE) = ''
    OR DEPARTAMENTO IS NULL OR TRIM(DEPARTAMENTO) = ''
    OR CATEGORIA IS NULL OR TRIM(CATEGORIA) = '';


-- =====================================================
-- VALIDACIÓN ADICIONAL DE CALIDAD DE DATOS
-- =====================================================


-- 13. RESUMEN DE CAMPOS INCOMPLETOS
-- Cada columna cuenta sus propios registros incompletos.
-- Un mismo registro puede aparecer en más de un conteo.
SELECT
    COUNT(*) AS total_registros,

    SUM(CASE
        WHEN COD_IPRESS IS NULL OR TRIM(COD_IPRESS) = ''
        THEN 1 ELSE 0
    END) AS cod_ipress_incompleto,

    SUM(CASE
        WHEN NOMBRE IS NULL OR TRIM(NOMBRE) = ''
        THEN 1 ELSE 0
    END) AS nombre_incompleto,

    SUM(CASE
        WHEN DEPARTAMENTO IS NULL OR TRIM(DEPARTAMENTO) = ''
        THEN 1 ELSE 0
    END) AS departamento_incompleto,

    SUM(CASE
        WHEN CATEGORIA IS NULL OR TRIM(CATEGORIA) = ''
        THEN 1 ELSE 0
    END) AS categoria_incompleta

FROM fia_databricks.stasys_team02_alejandro.renipress_silver;


-- 14. VERIFICAR CÓDIGOS IPRESS NULOS O VACÍOS
SELECT
    COUNT(*) AS codigos_ipress_incompletos
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE
    COD_IPRESS IS NULL
    OR TRIM(COD_IPRESS) = '';


-- 15. DISTRIBUCIÓN DE IPRESS POR INSTITUCIÓN
SELECT
    COALESCE(INSTITUCION, 'SIN INSTITUCION') AS institucion,
    COUNT(*) AS total_ipress
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
GROUP BY COALESCE(INSTITUCION, 'SIN INSTITUCION')
ORDER BY total_ipress DESC;


-- 16. TOP 10 DEPARTAMENTOS CON MÁS IPRESS
SELECT
    DEPARTAMENTO,
    COUNT(*) AS total_ipress
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE DEPARTAMENTO IS NOT NULL
  AND TRIM(DEPARTAMENTO) <> ''
GROUP BY DEPARTAMENTO
ORDER BY total_ipress DESC
LIMIT 10;


-- 17. VERIFICAR VALORES DE CATEGORÍA NULOS O VACÍOS
SELECT
    COUNT(*) AS categorias_incompletas
FROM fia_databricks.stasys_team02_alejandro.renipress_silver
WHERE
    CATEGORIA IS NULL
    OR TRIM(CATEGORIA) = '';


-- =====================================================
-- FIN DE LAS CONSULTAS SQL - PROYECTO STASYS
-- =====================================================
```
